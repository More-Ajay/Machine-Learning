# Decision Tree

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/content/Company_Data.csv')
data
```

```python
# find missing value

data.isnull().sum()
```

```python
# find duplicate value

data.duplicated().sum()
```

```python
# dataset information

data.info()
```

```python
# statistical summary

data.describe()
```

```python
data.drop(columns= 'US' , axis=1 , inplace=True)
```

```python
data
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data['ShelveLoc'] = le.fit_transform(data['ShelveLoc'])
data['Urban'] = le.fit_transform(data['Urban'])
```

```python
data
```

```python
# features & target

x = data.iloc[:,0:9].values
y = data['Urban']
```

```python
# split the data

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2)
```

```python
# model train

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()

model.fit(x_train , y_train)
```

```python
# prediction

y_pred = model.predict(x_test)
y_pred
```

```python
# accuracy

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test , y_pred)
accuracy
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(model , filled=True)
```

# Random Forest

```python
data = pd.DataFrame({
    'Age': [25, 35, 45, 20, 55, 60, 22, 40, 30, 48],
    'Income': [50, 90, 110, 30, 70, 120, 45, 85, 40, 95],
    'Bought': [0, 1, 1, 0, 1, 1, 0, 1, 0, 1]
})

data
```

```python
# dataset information

data.info()
```

```python
# statistical summary

data.describe()
```

```python
# find missing value

data.isnull().sum()
```

```python
# find duplicates value

data.duplicated().sum()
```

```python
# features & target

x = data[['Age' , 'Income']]
y = data['Bought']
```

```python
# split the data

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2)
```

```python
# train model

from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=5)

model.fit(x_train , y_train)
```

```python
# prediction

y_pred = model.predict(x_test)
y_pred
```

```python
# prediction on new data

yy_pred = model.predict([[40,100]])
yy_pred
```

```python
# accuracy

from sklearn.metrics import accuracy_score

score = accuracy_score(y_pred , y_test)
score
```

# K-Means

```python
df = pd.DataFrame({
    "Age":[22,25,27,40,42,45],
    "Spending":[15000,18000,20000,85000,90000,95000]
})

df
```

```python
# dataset information

df.info()
```

```python
# statistical summary

df.describe()
```

```python
# find missing value

df.isnull().sum()
```

```python
# find duplicate values

df.duplicated().sum()
```

```python
# features & target

x = df[['Age']]
y = df['Spending']
```

```python
# split the data

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2)
```

```python
# train model

from sklearn.cluster import KMeans

model = KMeans(n_clusters=2)

model.fit(x_train , y_train)
```

```python
# prediction

pred = model.predict(x_test)
pred
```

```python
# cluster centroids

centroids = model.cluster_centers_
```

```python
df['Cluster'] = model.fit_predict(df)
```

```python
df
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x , y , label='Actual Value')
plt.scatter(centroids[:, 0], centroids[:, 1], c="red" , label="Centroids")
plt.title('Age Vs Spending')
plt.xlabel('Age')
plt.ylabel('Spending')
plt.legend()
plt.show()
```

# DBSCAN

```python
data = pd.DataFrame({
    "X":[1,2,2,3,8,8,25],
    "Y":[1,2,1,2,8,9,30]
})

data
```

```python
# dataset information

data.info()
```

```python
# statisctical summary

data.describe()
```

```python
# find missing value

data.isnull().sum()
```

```python
# find duplicated value

data.duplicated().sum()
```

```python
# feature & target

x = data[['X']]
y = data['Y']
```

```python
# split the data

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2)
```

```python
# train model

from sklearn.cluster import DBSCAN

model = DBSCAN(eps=3 , min_samples=2)

model.fit(x_train , y_train)
```

```python
# prediction

prediction = model.fit_predict(x_test)
prediction
```

```python
data['Cluster'] = model.fit_predict(data)
```

```python
data
```

```python
plt.scatter(x , y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("DBSCAN Clustering")

plt.show()
```
